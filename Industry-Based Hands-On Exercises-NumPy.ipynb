# Hands-On Esxercise 1 - Business Anlaytics (Sales Performance Calculator)

Using NumPy to perform basic numerical analysis on the sales data of a growing retail business . The management team wants to understand the week's peformance.

# Task 1

```python
# Import NumPY

import numpy as np

# create the daily_sales array
daily_sales = np.array([125000, 150000, 175000,140000, 190000, 210000, 160000])

# print the array
print("Daily Sales:", daily_sales)

# Print the type of the array
print("Daily Sales Type:", daily_sales.dtype)

# Shape of the array

print("Daily Sales Shape:", daily_sales.shape)
```

# Task 2

```python
# Calculate the total sales. To get total sales, the sum of the daily sales is needed

total_sales = np.sum(daily_sales)

# Print result
print("Total Sales:", total_sales)
```

# Task 3

```python
# Calculate Average Daily Sales

average_sales = np.sum(daily_sales)/ len(daily_sales)

# Print result
print("Average Sales:", average_sales)
```

# Task 4

```python
# Applying a sales adjustment. 

daily_sales = np.array([125000, 150000, 175000,140000, 190000, 210000, 160000])

# Rule: Each sales get 10% increase of the original price i.e original price + 10% increase
# A 10% increase = 100 + (10/100) = 1.10 
increase = 1.10

# create a new array to represent the adjusted sales by multiply each day sales by 1.10 

adjusted_sales = np.multiply(daily_sales, increase)

# Print result

print("Adjusted Sales:", adjusted_sales)
```

# Task 5

```python
# Compare Original sales and the adjusted sales by calculting the additional revenue from 10% sales increase

daily_sales = np.array([125000, 150000, 175000,140000, 190000, 210000, 160000])

sales_difference = np.subtract(adjusted_sales, daily_sales)

# print result

print("Sales Difference:", sales_difference)
```

# Task 6

```python
# Using NumPy Mathematical Functions to calculate the average sales.

average_sales = np.average(daily_sales)

# Print results
print("Total Sales:", total_sales)
print("Average Sales:", average_sales)
```

Explanation: The total sales show the combined revenue generated over the seven-day period, while the average dail sales shows the typical amount generated per day. The average makes it easier for the usiness to understand its daily sales performance an compare it with the performance of the other periods.

# Hands- On Exercise 2 - Education Analytics (Student Performance Analysis)

Using NumPy to perform numerical analysis for a training program that recorded the scores of a group of students . The program coordinator wants to understand the overall performance of the students and how individual scores differ from the group average.

# Task 1

```python
# import NumPy as np

import numpy as np

# create the numpy array array

scores = np.array([62, 75,81,69,88,94,73,85,77,91])

# Print result
print("Student Scores:", scores)
print("Score Type:", scores.dtype)
print("Shape:", scores.shape)
```

# Task 2

```python
# Calculate the mean score

mean_score = np.mean(scores)

# Print result
print("Mean Score:", mean_score)
```

# Task 3

```python
# Calculate each student deviation.How far is each student's score from the average(mean)

scores = np.array([62, 75,81,69,88,94,73,85,77,91])
mean_score = np.mean(scores)

deviations = np.subtract(scores, mean_score)

# Print result

print("Deviations:", deviations)
```

# Task 4

Observations from the Deviations:

Positive deviation, negative derivation  and deviation close to zero.

This means a positive deviation means that a student's score is above the mean, while a negative deviation means that the score is below the group mean. A deviation close to zero means that te student's score is close to the average score of the group.


# Task 5

```python
# Create a new aray contatining the squared deviations.

scores = np.array([62, 75,81,69,88,94,73,85,77,91])
mean_score = np.mean(scores)

deviations = np.subtract(scores, mean_score)

# Squaring the deviations
squared_deviations = deviations ** 2

# Print result
print("Squared Deviations:", squared_deviations)
```

The deviations are squared so that all the values become positive. This prevents positive and negative deviations from cancelling each other out and allows us to measure the size of the differences from the mean.

# Task 6

```python
# Calculate the sum of squared deviations

squared_deviations = deviations ** 2

sum_squared_deviations = np.sum(squared_deviations)

# Print result

print("Sum of Squared Deviations:", sum_squared_deviations)
```

# Task 7

## Connecting the steps to Standard Deviation

The standard deviation calculation involves several steps:

1. Calculate the mean of the scores.

2. Calculate the deviation of each score from the mean

3. Square each deviation.

4. Add the squared deviations together.

5. Divide the sum by the appropriate number of observations

6. Take the square root of the result

Standard deviation is useful because it shows how spread out the student scores are around the mean. While the mean score tells us the average performance, standard deviation gives additional information about how consistently the students performed. A smaller standard deviation indicates that the scores  are closer together, while a larger standard deviation indicates greater variation is performance.

Mean score = 79.5 ---> Deviations -----> squared deviations ----> sum of squared deviations = 932.5

# Hands-On 3- Engineering & Scientific Computing (Trigonometric Series and Convergence

Using NumPy to create a sequence of values, work wirh a trigonometric function, and observe observe the idea of convergence. The team wants to understand how an approximation changes as more terms are included in a calculaion.


# Task 1

```python
#import numpy as np

import numpy as np

# Define the angle

theta = 30

# convert the angle to radians

theta_radians = np.radians(30)

# print result
print("Angle:", theta_radians)
```

# Task 2

```python
# Create a sequence of values for k. 

k = np.arange (1, 100, 1)

# print the resulting array for the sequence. Fist value, last value and the number of values generated

print("K Sequence:",k)
print("First Value:", k[0])
print("Last value:", k[-1])
print("Number of values:", len(k))
```

# Task 3

```python
#calculate the trigonometric component

sin_theta = np.sin(theta_radians)

# print result

print("Sin Theta:", sin_theta)
```

# Task 4

```python
# Building the series terms

sin_theta = np.sin(theta_radians)
k = np.arange (1, 100, 1)

# create the individual terms of the series by diving the sine component by k

terms = np.divide(sin_theta, k)

#print result

print("Terms:", terms)
```

# Task 5

```python
# Calculate the sum

series_sum = np.sum(terms)

#print result

print(series_sum)
```

# Task 6

```python
#Increase the number of terms

# For 99 terms
k_99 = np.arange(1, 100, 1)
terms_99 = np.divide(sin_theta, k_99)
sum_99 = np.sum(terms_99)

# For 999 terms
k_999 = np.arange(1, 1000, 1)
terms_999 = np.divide(sin_theta, k_999)
sum_999 = np.sum(terms_999)

# For 9999 terms
k_9999 = np.arange(1, 10000, 1)
terms_9999 = np.divide(sin_theta, k_9999)
sum_9999 = np.sum(terms_9999)

# print result
print("99 terms:", sum_99)
print("999 terms:", sum_999)
print("9999 terms:", sum_9999)
```

# Task 7

# Investigate convergence

Term (99) --------->>>> 2.5886887588198104

Term(999) ---------->>>>> 3.7422354302751724

Term(9999) --------->>>>>> 4.89375301802219

As the number of terms increases, the series result also increases. The additional terms contribute smaller values to the total, so the result changes progressively as more terms are included


# Task 8

# Convergence Explanation wth the experiment

Convergence means that as more terms are added, the result  approaches a limiting value. The additional terms become smaller, so their effect on the total gradually decreases. When the result gets closer to a limiting value, t indicates that the series is approaching convergence. 

Basically, when thr fewer terms are used in the series, the rsult is based on a smaller number of vales and gives a lower approximation. As the number of terms increases, more values are included in the calculation, causing the series result to change and ecome a better approximation of the series.

```python
print("="* 50)
print()
print("    ENERGY CONSUMPTION ANALYSIS")
print("="*50)
print(f""" Scenario:

An energy analyst wants to examine the electricity consumption of a small facility over seven days.

Objective:

The objective is to use NumPy to calculate the total and average electricity consumption, estimate the variation in daily consumption, 
and examine how consumption changes across the seven-day period.
""")

print()

# Data Set + Code

# import numpy as np

import numpy as np

# Daily electricity consumption in kWh

consumption = np.array([120, 150, 135, 180, 160,200,175])

# Print result

print("Daily Consumption:", consumption)
print()
# calculate the total and average

total_consumption = np.sum(consumption)
average_consumption = np.average(consumption)

# Print result

print("Total Consumption:", total_consumption, "kWh")
print("Average Daily Consumption:", average_consumption, "kWh")
print()
#calculate for the devation

deviation = np.subtract(consumption, average_consumption)

# Print result

print("Deviation from Average:", deviation)
print()
# using np.sqrt()

squared_deviation = deviation** 2
mean_squared_deviation = np. mean(squared_deviation)
standard_deviation = np.sqrt(mean_squared_deviation)

# print result
print("Standard deviation:", standard_deviation, "kWh")
print()


print(F""" RESULTS AND INTERPRETATION
    Results
    
The total electricity consumption over the seven-day period was 1,120kWh, while the average daily consumption was 160kWh. 
The standard deviation was approximaely 25.49kWh.

    Interpretation

The facility consumed an average of 160kWh per day. Some days recorded consumption above the average , while others were below it. 
The standard deviation of approximately 25.49 kWh shows that daily electricity consumption w=varied around the average rather than remaining constant.

""")

print()
print(F""" CONCLUSION

This experiment demonstrates how NumPy can be used to analyse real-word energy consumpton data.
NumPy functions such as np.array(),np.sum(), and np.sqrt() made it possible to calculate and interpret important measures of electricity consumption 
efficiently.
""")

print("="*50)
```
